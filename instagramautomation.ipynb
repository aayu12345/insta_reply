```python
import requests
import json

# References:
# https://developers.facebook.com/docs/instagram-platform/instagram-api-with-facebook-login/comment-moderation
# https://developers.facebook.com/docs/instagram-platform/instagram-graph-api/reference/ig-media/comments
# https://developers.facebook.com/docs/instagram-platform/instagram-graph-api/reference/ig-comment/replies#create
# https://developers.facebook.com/docs/instagram-platform/reference/instagram-media#edges
# https://developers.facebook.com/docs/instagram-platform/instagram-graph-api/reference/ig-user/media#reading

# Get Comments & Its Replies

business_discovery_parameters = "thumbnail_url,media_type,media_product_type,timestamp,like_count,comments_count,media_url,permalink"

other_parameters = "comments{id,from,text,timestamp,like_count,media,parent_id,replies{from,timestamp,like_count,text},hidden,user,username}"

fields = business_discovery_parameters + "," + other_parameters

url = f"https://graph.facebook.com/v21.0/{ig_user_id}/media"

payload = {
    "fields": fields,
    "access_token": long_access_token
}

response = requests.get(url, params=payload)
data = response.json()

print(json.dumps(data, indent=4))

# Extract comments from second media item
image_comment = data["data"][1]["comments"]["data"]

print(json.dumps(image_comment, indent=4))
```

```python
import requests
import json

# -------------------- ADD COMMENT --------------------

# Get media_id from fetched data
media_id = data["data"][1]["id"]

print(media_id)

# API to add comment
url = f"https://graph.facebook.com/v21.0/{media_id}/comments"

payload = {
    "message": "It's my first post!",
    "access_token": long_access_token
}

response = requests.post(url, params=payload)
comment_response = response.json()

print(json.dumps(comment_response, indent=4))


# -------------------- REPLY TO COMMENT --------------------

# Get comment_id from fetched comments
comment_id = data["data"][1]["comments"]["data"][0]["id"]

print(comment_id)

# API to reply to comment
url = f"https://graph.facebook.com/v21.0/{comment_id}/replies"

payload = {
    "message": "@_mountains1223 Thanks ❤️",
    "access_token": long_access_token
}

response = requests.post(url, params=payload)
reply_response = response.json()

print(json.dumps(reply_response, indent=4))
```

```python
import requests
import json

# Step 1: Get all media (posts)
media_url = f"https://graph.facebook.com/v21.0/{ig_user_id}/media"

media_params = {
    "fields": "id,comments{id,text}",
    "access_token": long_access_token
}

media_res = requests.get(media_url, params=media_params)
media_data = media_res.json()

# Step 2: Loop through posts
for media in media_data.get("data", []):

    media_id = media["id"]
    comments = media.get("comments", {}).get("data", [])

    # Step 3: Loop through comments
    for comment in comments:
        comment_id = comment["id"]
        comment_text = comment["text"]

        print(f"Replying to comment: {comment_text}")

        # Step 4: Reply to comment
        reply_url = f"https://graph.facebook.com/v21.0/{comment_id}/replies"

        payload = {
            "message": "Thanks 😊",
            "access_token": long_access_token
        }

        response = requests.post(reply_url, params=payload)
        reply_data = response.json()

        print("Reply Response:", json.dumps(reply_data, indent=4))
```
